#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python
# Set vacío donde guardaré los nombres de los aprobados
aprobados = set()

# Recorro cada usuario (cada diccionario de la lista):
for usuario in datos:
    usuario["nombre"] = usuario["nombre"].capitalize()    # Primera letra del nombre en mayúscula
    usuario["total"] = sum(usuario["puntos"])             # Suma de los puntos, guardada en la nueva clave "total"
    usuario["aprobado"] = usuario["total"] > 50           # Booleano: True si el total es mayor que 50
    
    if usuario["aprobado"]:                               # Si está aprobado, 
        aprobados.add(usuario["nombre"])                  # añado su nombre al set
 
print("\033[1mDatos:\033[0m", datos)
print("\033[1mAprobados:\033[0m", aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
# Creo un diccionario vacío para guardar los resultados:
diccionario_estudiantes = {}

# Recorro cada estudiante (me quedo solo con el diccionario interno)
for datos in estudiantes.values():
    nombre = datos["nombre"].upper()        # Nombre en mayúsculas, que será la clave
    notas = datos["notas"]                  # Lista de notas del estudiante
    promedio = sum(notas) / len(notas)      # Promedio = suma de notas entre cantidad de notas
    
    # Creo la entrada con la estructura:
    diccionario_estudiantes[nombre] = {                  
    "promedio": round(promedio,1),          # Promedio de las notas con 1 decimal
    "aprobado": promedio >= 20,             # True si aprueba (promedio>=20) y False si no
    "max_min": (max(notas), min(notas))     # Tupla con nota máxima y mínima
    }

diccionario_estudiantes
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
#Creamos con open y w+. Escribimos con la función write y cerramos. 
archivo = open("notas_alumnos.txt", "w")
archivo.write("Ana,10,20,30\nLuis,5,15\nMarta,100,50,25\n")
archivo.close()
```

```python
#Creamos el diccionario notas
resultado = []                                                    #creamos una lista vacía para meter los diccionarios
with open("notas_alumnos.txt", "r") as archivo:                   #abrimos el archivo con la función r para leer
    for linea in archivo:                                         #recorremos cada linea del archivo
        partes = linea.split(",")                                 #partimos el texto por las comas
        nombre = partes[0].capitalize()                           #ponemos la primera posición en mayúscula
        notas = [] 
        for n in partes[1:]:                                      #empezando desde la posición 1
            notas.append(int(n))                                  #convertimos los números de texto a nums para poder sumarlos
        promedio = sum(notas) / len(notas)                        #sumamos las notas y las dividimos entre las que hay
        if promedio >= 20:                                        #bucle if para ver si está aprobado o no
            aprobado = True
        else:
            aprobado = False

        resultado.append({                                        #creamos el diccionario para guardar los resultados
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        })

with open("resultado.json", "w", encoding="utf-8") as salida:
    json.dump(resultado, salida)
        
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
#creamos el diccionario con la key datos y el valor es la lista con un dic por persona
datos = {
    "usuarios": [
        {"id": 1, "nombre": "Ana", "roles": ["user", "admin"]},
        {"id": 2, "nombre": "Luis", "roles": ["user"]},
        {"id": 3, "nombre": "Marta", "roles": ["moderator", "user"]},
    ]
}

#creamos el archivo con open y escribimos con la w. Pasamos a formato json
with open("usuarios.json", "w") as archivo:
    json.dump(datos, archivo)

#Leemos el archivo con r. Por cada usuario en datos, pasarlo a mayúscula con upper
with open("usuarios.json", "r") as archivo:
    for usuario in datos["usuarios"]:
        usuario["nombre"] = usuario["nombre"].upper()

# crear usuarios.csv
with open("usuarios.csv", "w") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["id", "nombre", "roles"])
    for usuario in datos["usuarios"]:
        escritor.writerow([usuario["id"], usuario["nombre"], "|".join(usuario["roles"])])
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
f = open("fragmento-libro.txt", "r")     # Abrimos el archivo en modo lectura
texto = f.read().lower()                 # Leemos el archivo, ignorando mayúsculas y minúsculas
f.close()

palabras = re.findall(r"\w+", texto)     # Ignoramos los signos, nos quedamos sólo con las palabras

frecuencias = {}
for palabra in palabras:
    if palabra in frecuencias:
        frecuencias[palabra] += 1        # Suma 1 si la palabra vuelve a salir, y guarda el valor en el diccionario creado
    else:
        frecuencias[palabra] = 1

top5 = dict(sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)[:5])
        # sorted ordena según las keys (alfabéticamente), de menor a mayor
        # key = lambda x: x[1] ordena según los valores de las llaves (la frecuencia de las palabras)
        # reverse = True hace que primero vayan los valores más altos (palabras mása frecuentes)
        # [:5] nos quedamos con los 5 primeros elementos de la lista ordenada

f = open("frecuencias.json", "w")
json.dump(top5, f, indent=4)
f.close()

print(top5)
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
#creamos los tres archivos con el argumennto w para escribir
with open("ventas1.csv", "w", encoding="utf-8") as archivo:
    archivo.write("producto,cantidad\nA,10\nB,5\n")

with open("ventas2.csv", "w", encoding="utf-8") as archivo:
    archivo.write("producto,cantidad\nA,10\nB,5\n")

with open("ventas3.csv", "w", encoding="utf-8") as archivo:
    archivo.write("producto,cantidad\nA,10\nB,5\n")

#lo leemos y los sumamos por producto
totales = {}                                                                #creamos un dic vacio para ir guardando los producots
for variable in range(1, 4):                                                #hacemos un bucle que se repite 3 veces
    with open(f"ventas{variable}.csv", "r", encoding="utf-8") as archivo:   #abrimos el archivo y lo leemos con el argumento r
        lineas = archivo.readlines()                                        #lee el archivo y lodevuleve como un lista
    for linea in lineas[1:]:                                                #por cada linea a partir de la posición 1
        partes = linea.split(",")                                           #cortamos la lista por las comas
        producto = partes[0]                                                #las posiciones 0 son los productos
        cantidad = int(partes[1])                                           #las posiciones 1 son las cantidades, que pasamos de texto a número para poder sumarla
        totales[producto] = totales.get(producto, 0) + cantidad             #si el producto no está en el dic, devuelve 0 y súmale la cantidad
        
#guardamos en el archivo total_ventas.csv.
with open("total_ventas.csv", "w", encoding="utf-8") as archivo:
    archivo.write("producto,cantidad\n")
    for producto in totales:
        archivo.write(f"{producto},{totales[producto]}\n")

print(totales)
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
#creamos alumnos.json 
alumnos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]},
]

with open("alumnos.json", "w", encoding="utf-8") as archivo:
    json.dump(alumnos, archivo)         

# creamos el diccionario vacio y leemos el archivo con el argumento r
notas_finales = {}                                                   
with open("alumnos.json", "r", encoding="utf-8") as archivo:     
    datos = json.load(archivo)                                

for alumno in datos:                                             #por cada alumno en datos 
    nombre = alumno["nombre"]                                    #sacamos el nombre
    notas = alumno["notas"]                                      #sacamos notas
    promedio = sum(notas) / len(notas)                           #sumamos las notas y dividimos entre las que hay

    if promedio >= 20:                                           #si el promedio es igual o mayor que 20 está aprobado, si no, false
        aprobado = True
    else:
        aprobado = False

    notas_finales[nombre] = {                                   #metemos el otro diccionario como valor en este diccionario
        "promedio": promedio,
        "max": max(notas),                                       #nota más alta
        "min": min(notas),                                       #nota más baja
        "aprobado": aprobado
    }

notas_finales
```
