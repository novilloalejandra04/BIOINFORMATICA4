### ARCHIVOS

Python emplea unos file objects para interactuar con los archivos de tu ordenador.

Puede manejar cualquier tipo de archivo, para algunos es posible que necesites librerías adicionales.

Podemos abrir archivos con **open()**, leerlos con **read() o readlines()**, cerrarlos con **close()** y escribir en ellos con **write()**. 

Usando **open()** y w, w+, a+ nos permite sobreescribir o añadir información a la ya existente.

```python
# Vamos a abrir el archivo test.txt. esta dentro de la carpeta que nos descargamos del campus. MEjor que esté en la misma carpeta., porqie si no hay qque escribir la ruta entera
mi_archivo=open('test.txt')
```

```python
mi_archivo
#Nos indica que hay un archivo pero no te lo lee. Input/Output
```

```python
# Leemos el archivo. 
mi_archivo.read()
```

```python
# Pero, qué pasa si lo leemos más de una vez? Que nos lleva el cursor al final. Es decir, ya lo ha leído y el cursor no tiene nada más
mi_archivo.read()
```

```python
# Podemos reseatear el cursor así, lo pones al principio
mi_archivo.seek(0)
```

```python
mi_archivo.read()
```

```python
mi_archivo
```

```python
# Si usamos la función readlines() nos devuelve una lista para cada línea del archivo. Readlines te guarda el texto en listas, va por.
mi_archivo.seek(0)
mi_archivo.readlines()
```

```python
# Podemos cerrar un archivo. Siempre es buena idea cerrarlos para evitar que consuman memoria.
mi_archivo.close()
```

```python
mi_archivo=open("test.txt")
```

```python
contenido=mi_archivo.readlines()
```

```python
contenido
```

```python
# Podemos escribir un archivo usando open y w+ que se emplea para leer y escribir sobre el archivo. Open por defecto abre, pero tiene otras funciones, para que las haga, le añades una argumento que se llama W+.
# Hay que tener en cuenta que si se usa w o w+ se elimina el contenido anterior del archivo!!!!. En este caso te abre, lee, borra y sobreescribe
mi_archivo=open('test.txt','w+')
```

```python
# Ahora escribimos en él. Para decirte que sí ha funcionado te sale un  num,en este caso el 23
mi_archivo.write('Esta es una linea nueva')
```

```python
mi_archivo.seek(0)
mi_archivo.read()
```

```python
mi_archivo.close()
```

```python
# Ahora queremos añadir información nueva
mi_archivo = open('test.txt','a+')
mi_archivo.write('\nEsta linea va detras de la anterior')
mi_archivo.write('\nY aqui ponemos otra')
```

```python
mi_archivo.seek(0)
mi_archivo.read()
# La barra invertiva es para bajar renglón, python lo pone eo, si quieres que lo ponga mono le da<s a print.
```

```python
mi_archivo.seek(0)
print(mi_archivo.read())
```

```python
# También podemos iterar sobre los archivos. Le pones el cursor al principio y le dices que para cada linea de tu archivo haga algo, en este caso print, mostarlo. Te lo hace tantas vecees como lineas haya. 
mi_archivo.seek(0)
for line in mi_archivo:
    print(line)
```

```python
mi_archivo.close()
```
